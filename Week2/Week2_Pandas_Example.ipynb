[심화] 불린 인덱싱

---
조건식을 [ ] 안에 기입하여 간편하게 데이터 필터링 적용


titanic_df[titanic_df['Age'] > 60][['Name','Age']].head(3) 코드 설명

---


이 코드는 titanic_df DataFrame에서 60세 초과인 승객들의 이름과 나이를 상위 3명만 추출하여 보여주는 코드입니다. 각 부분별로 설명해 드릴게요

* titanic_df['Age'] > 60: 이 부분은 titanic_df DataFrame의 'Age' 컬럼에서
나이가 60보다 큰(초과하는) 모든 행을 선택하는 불린 인덱싱(Boolean Indexing) 조건입니다. 결과는 True 또는 False로 이루어진 Series가 됩니다.
* titanic_df[...]: 위에 설명된 불린 Series를 titanic_df 뒤의 대괄호 안에 넣으면, True로 표시된 행들만 필터링하여 새로운 DataFrame을 만듭니다. 즉, 나이가 60세 초과인 승객들의 모든 정보가 담긴 DataFrame이 됩니다.
* [['Name','Age']]: 필터링된 DataFrame에서 'Name' 컬럼과 'Age' 컬럼만을 선택합니다. 대괄호 두 개를 사용한 것은 여러 개의 컬럼을 선택하겠다는 의미이며, 결과는 DataFrame 형태입니다.
* head(3): 마지막으로, 앞에서 선택된 이름과 나이 컬럼을 가진 DataFrame의 상위 3개 행만 반환하여 출력합니다.

```python
titanic_df[titanic_df['Age'] > 60][['Name','Age']].head(3)
```

**데이터 그룹핑과 필터링**


---
* 울릉도 날씨 데이터 다시 활용
* weather 이름의 DataFrame 활용 및 불러오기


```python
import pandas as pd

path = 'https://github.com/yjun13/ML_2026_Spring/raw/main/Data/'
weather_file = path + 'weather.csv'

weather = pd.read_csv(weather_file, encoding='CP949')
print(weather.head(3))
print('weather 데이터의 shape :', weather.shape)
```

```python
weather['month'] = pd.DatetimeIndex(weather['일시']).month
# '일시' 컬럼은 문자열이므로 평균 계산에서 제외하고, 숫자형 컬럼에 대해서만 평균을 계산합니다.
means = weather.groupby('month')[['평균기온', '최대풍속', '평균풍속']].mean()
print(means)
```

```python
means['평균풍속']>=4.0
```

```python
means[ means['평균풍속'] >= 4.0 ]
```

데이터 구조의 변경: concat

```python
df_1 = pd.DataFrame( {'A' : ['a10', 'a11', 'a12'],
                      'B' : ['b10', 'b11', 'b12'],
                      'C' : ['c10', 'c11', 'c12']} ,
                      index = ['가', '나',  '다'] )

df_2 = pd.DataFrame( {'B' : ['b23', 'b24', 'b25'],
                      'C' : ['c23', 'c24', 'c25'],
                      'D' : ['d23', 'd24', 'd25']} ,
                      index = ['다', '라',  '마'] )
```

```python
df_3 = pd.concat( [df_1, df_2], axis=0)
print(df_3)
```

```python
df_4 = pd.concat( [df_1, df_2], axis=1)
print(df_4)
```
